```python
# -*- coding: utf-8 -*-
# ==========================================================================
#  SmartIE – Sistema Inteligente de Apoyo a Decisiones para Ingeniería Industrial
#  Mini-ERP / Mini-Oracle: Inventario + Pedidos + Proveedores + OEE + What-If
#  Compatible con Google Colab (pegar completo en UNA celda y ejecutar)
# ==========================================================================
#
#  MÓDULOS
#  1. Inventario avanzado ..... EOQ, ROP, stock de seguridad (variabilidad de
#                               demanda y de lead time), ABC, cobertura, alertas
#                               y costo total de inventario.
#  2. Pedidos de clientes ..... urgencia, disponibilidad de stock (con reparto
#                               por prioridad) y pedidos en riesgo.
#  3. Proveedores y riesgos ... confiabilidad, riesgo Alto/Medio/Bajo, lead
#                               time efectivo y recomendaciones.
#  4. OEE ..................... disponibilidad, rendimiento, calidad, pérdidas.
#  5. Costos y What-If ........ costo de mantener vs. faltantes y simulación de
#                               retrasos de proveedores.
#  6. Dashboard ............... 6 gráficos principales + gráficos de escenarios.
#  7. Resumen ejecutivo ....... conclusiones y acciones prioritarias.
#  Extra: optimización del mix de producción con PuLP y exportación a Excel.
# ==========================================================================

# ==========================================================================
# 0. CONFIGURACIÓN E IMPORTACIONES
# ==========================================================================
import sys, subprocess, importlib, re, warnings
from datetime import datetime, timedelta

def _asegurar(paquete):
    """Instala el paquete si no existe (útil en Colab)."""
    try:
        importlib.import_module(paquete)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", paquete])

_asegurar("pulp")
_asegurar("scipy")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
from pulp import LpProblem, LpMaximize, LpVariable, lpSum, value, PULP_CBC_CMD
try:                                   # Versiones nuevas de PuLP ya no exportan LpStatus en la raíz
    from pulp import LpStatus
except ImportError:
    try:
        from pulp.constants import LpStatus
    except ImportError:
        LpStatus = {1: "Optimal", 0: "Not Solved", -1: "Infeasible", -2: "Unbounded", -3: "Undefined"}

# ---- Empresa ficticia usada SOLO para la demostración ---------------------
EMPRESA_EJEMPLO = "Manufacturas Industriales del Bajío S.A. de C.V."
PERFIL_EMPRESA = ("Empresa ficticia de León, Gto., que ensambla bombas hidráulicas, paneles de control y soportes "
                  "metálicos reforzados para clientes industriales y de construcción. Opera un turno de 8 h en su "
                  "línea principal y compra materiales a 5 proveedores nacionales.")

warnings.filterwarnings("ignore")
print("✅ SmartIE v2.1 cargado (importación de PuLP corregida)")
sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.dpi"] = 100

try:
    from IPython.display import display as _display
except ImportError:
    _display = print

# ---- Parámetros globales (ajustables) ------------------------------------
DIAS_ANIO = 365
FACTOR_COSTO_FALTANTE = 0.40   # Costo de faltante = 40% del costo unitario (margen perdido + penalización)
Z_RIESGO_LT = 1.645            # Lead time "pesimista" = LT + 1.645·σ (≈95%)
UMBRAL_URGENTE = 5             # días restantes <= 5  -> Urgente
UMBRAL_PROXIMO = 12            # días restantes <= 12 -> Próximo
HORIZONTE_PROD_DIAS = 5        # Horizonte del plan de producción (días)
MODO_DEMO = False              # Si es True, no pregunta nada y usa datos de ejemplo

COL_ESTADO = {"Cubierto": "#2ecc71", "Cubrible con reposición": "#f1c40f",
              "EN RIESGO": "#e74c3c", "Sin datos": "#95a5a6"}

# ==========================================================================
# 1. UTILIDADES DE INTERFAZ
# ==========================================================================
def titulo(texto, ancho=72):
    print("\n" + "=" * ancho)
    print(texto)
    print("=" * ancho)

def mostrar(df):
    _display(df)

def preguntar_si_no(mensaje):
    """Pregunta s/n. En modo demo responde 's' automáticamente."""
    if MODO_DEMO:
        print(f"{mensaje} → s (modo demo)")
        return True
    return input(f"{mensaje} (s/n): ").strip().lower() == "s"

def pedir_texto(mensaje, defecto=None):
    if MODO_DEMO and defecto is not None:
        return defecto
    txt = input(f"{mensaje}" + (f" [{defecto}]" if defecto is not None else "") + ": ").strip()
    return txt if txt else defecto

def pedir_float(mensaje, defecto=None, minimo=0.0):
    while True:
        txt = input(f"{mensaje}" + (f" [{defecto}]" if defecto is not None else "") + ": ").strip()
        if txt == "" and defecto is not None:
            return float(defecto)
        try:
            v = float(txt)
            if v < minimo:
                print(f"   ❌ El valor debe ser >= {minimo}.")
                continue
            return v
        except ValueError:
            print("   ❌ Ingresa un número válido.")

def _prop(x):
    """Convierte 0.82 o 82 (%) a proporción 0-1."""
    x = float(x)
    return x / 100 if x > 1 else x

def _a_fecha(f):
    if isinstance(f, pd.Timestamp):
        return f.date()
    if isinstance(f, datetime):
        return f.date()
    return f

def _fecha(txt):
    """Acepta 'hoy', AAAA-MM-DD o un entero relativo (+12 / -3 días)."""
    txt = txt.strip().lower()
    hoy = datetime.now().date()
    if txt in ("hoy", ""):
        return hoy
    if re.fullmatch(r"[+-]?\d+", txt):
        return hoy + timedelta(days=int(txt))
    return datetime.strptime(txt, "%Y-%m-%d").date()

def _lista_materiales(texto):
    return [m.strip() for m in re.split(r"[;,]", str(texto)) if m.strip()]

def _leer_registros(instruccion, parser):
    """Lee líneas hasta que el usuario escriba 'fin'. Devuelve lista de dicts."""
    print(instruccion)
    print("   (escribe 'fin' para terminar)")
    registros = []
    while True:
        linea = input("→ ").strip()
        if linea.lower() == "fin":
            break
        try:
            registros.append(parser(linea))
            print(f"   ✔ Registro {len(registros)} agregado.")
        except Exception as e:
            print(f"   ❌ Formato inválido ({e}). Intenta de nuevo.")
    return registros

# ==========================================================================
# 2. CARGA DE DATOS (ejemplo o ingreso propio)
# ==========================================================================
def datos_inventario_ejemplo():
    return pd.DataFrame({
        "Material": ["Tornillo M8", "Placa Acero", "Motor 1.5HP", "Rodamiento", "Cable 3x12", "Válvula", "Sensor"],
        "Stock_Actual": [450, 80, 12, 95, 320, 28, 55],
        "Demanda_Diaria_Prom": [25, 4, 0.8, 6, 18, 1.5, 3],
        "Desv_Estandar_Demanda": [5, 1.2, 0.3, 1.5, 4, 0.4, 0.8],
        "Lead_Time_Dias": [7, 14, 21, 10, 5, 12, 8],
        "Costo_Pedido": [150, 800, 2500, 300, 200, 600, 400],
        "Costo_Mant_Unit_Anual": [2.5, 45, 180, 12, 3.5, 35, 15],
        "Costo_Unitario": [0.8, 120, 1850, 45, 8.5, 220, 95],
        "Nivel_Servicio_Deseado": [0.95, 0.98, 0.99, 0.95, 0.90, 0.97, 0.95],
    })

def cargar_inventario():
    titulo("📥 DATOS DE INVENTARIO / MATERIALES")
    if preguntar_si_no("¿Usar datos de ejemplo de inventario?"):
        print("✅ Inventario de ejemplo cargado.")
        return datos_inventario_ejemplo()

    def parser(linea):
        p = [x.strip() for x in linea.split(",")]
        if len(p) != 9:
            raise ValueError("se requieren 9 campos")
        return {"Material": p[0], "Stock_Actual": float(p[1]), "Demanda_Diaria_Prom": float(p[2]),
                "Desv_Estandar_Demanda": float(p[3]), "Lead_Time_Dias": float(p[4]),
                "Costo_Pedido": float(p[5]), "Costo_Mant_Unit_Anual": float(p[6]),
                "Costo_Unitario": float(p[7]), "Nivel_Servicio_Deseado": min(_prop(p[8]), 0.9999)}

    reg = _leer_registros(
        "Formato: Material, Stock, DemandaDiaria, DesvStdDemanda, LeadTime(días), CostoPedido, "
        "CostoMantUnitAnual, CostoUnitario, NivelServicio(0.95 ó 95)\n"
        "Ejemplo: Tornillo M8, 450, 25, 5, 7, 150, 2.5, 0.8, 0.95", parser)
    if not reg:
        print("⚠️ No ingresaste datos; se usará el ejemplo.")
        return datos_inventario_ejemplo()
    return pd.DataFrame(reg)

def datos_proveedores_ejemplo():
    return pd.DataFrame({
        "Proveedor": ["Aceros del Norte", "Motores Industrial", "Componentes SA", "Electrónica Plus", "Válvulas MX"],
        "Materiales_que_Suministra": ["Placa Acero", "Motor 1.5HP", "Tornillo M8; Rodamiento",
                                      "Sensor; Cable 3x12", "Válvula"],
        "Lead_Time_Promedio_Dias": [14, 21, 7, 8, 12],
        "Desviacion_Lead_Time": [3.5, 5.0, 1.5, 2.0, 3.0],
        "Porcentaje_Entregas_A_Tiempo": [0.82, 0.75, 0.93, 0.88, 0.79],
        "Retraso_Promedio_Dias": [2.8, 4.5, 0.8, 1.5, 3.2],
        "Costo_Unitario_Actual": [118, 1820, 0.78, 92, 215],
        "Calificacion": [7.5, 6.8, 9.1, 8.4, 7.0],
    })

def cargar_proveedores():
    titulo("📥 DATOS DE PROVEEDORES")
    if preguntar_si_no("¿Usar proveedores de ejemplo?"):
        print("✅ Proveedores de ejemplo cargados.")
        return datos_proveedores_ejemplo()

    def parser(linea):
        p = [x.strip() for x in linea.split(",")]
        if len(p) != 8:
            raise ValueError("se requieren 8 campos")
        return {"Proveedor": p[0], "Materiales_que_Suministra": p[1], "Lead_Time_Promedio_Dias": float(p[2]),
                "Desviacion_Lead_Time": float(p[3]), "Porcentaje_Entregas_A_Tiempo": _prop(p[4]),
                "Retraso_Promedio_Dias": float(p[5]), "Costo_Unitario_Actual": float(p[6]),
                "Calificacion": float(p[7])}

    reg = _leer_registros(
        "Formato: Proveedor, Materiales(separados por ;), LeadTimeProm, DesvLT, %ATiempo(0.82 ó 82), "
        "RetrasoProm, CostoUnit, Calificación(1-10)\n"
        "Ejemplo: Aceros del Norte, Placa Acero, 14, 3.5, 0.82, 2.8, 118, 7.5", parser)
    if not reg:
        print("⚠️ No ingresaste datos; se usará el ejemplo.")
        return datos_proveedores_ejemplo()
    return pd.DataFrame(reg)

def datos_pedidos_ejemplo():
    hoy = datetime.now().date()
    d = lambda n: hoy + timedelta(days=n)
    return pd.DataFrame({
        "Cliente": ["Empresa Alpha", "Constructora Beta", "Industria Gamma", "Taller Delta", "Empresa Alpha", "Industria Gamma"],
        "Material": ["Motor 1.5HP", "Placa Acero", "Válvula", "Tornillo M8", "Sensor", "Motor 1.5HP"],
        "Cantidad": [5, 40, 12, 800, 20, 10],
        "Fecha_Pedido": [d(-3), d(-5), d(-1), d(-7), d(-2), d(-1)],
        "Fecha_Requerida": [d(12), d(8), d(18), d(5), d(10), d(15)],
        "Prioridad": ["Alta", "Media", "Alta", "Baja", "Media", "Alta"],
    })

def cargar_pedidos():
    titulo("📥 PEDIDOS DE CLIENTES")
    if preguntar_si_no("¿Usar pedidos de ejemplo?"):
        print("✅ Pedidos de ejemplo cargados.")
        return datos_pedidos_ejemplo()

    def parser(linea):
        p = [x.strip() for x in linea.split(",")]
        if len(p) != 6:
            raise ValueError("se requieren 6 campos")
        prio = p[5].capitalize()
        if prio not in ("Alta", "Media", "Baja"):
            raise ValueError("prioridad debe ser Alta/Media/Baja")
        return {"Cliente": p[0], "Material": p[1], "Cantidad": float(p[2]),
                "Fecha_Pedido": _fecha(p[3]), "Fecha_Requerida": _fecha(p[4]), "Prioridad": prio}

    reg = _leer_registros(
        "Formato: Cliente, Material, Cantidad, FechaPedido, FechaRequerida, Prioridad\n"
        "Fechas: AAAA-MM-DD, 'hoy' o días relativos (-3, +12). Ejemplo: Empresa Alpha, Motor 1.5HP, 5, -3, +12, Alta", parser)
    if not reg:
        print("⚠️ No ingresaste datos; se usará el ejemplo.")
        return datos_pedidos_ejemplo()
    return pd.DataFrame(reg)

def datos_oee_ejemplo():
    return {"tiempo_planificado": 480, "tiempo_paros": 65, "total_producido": 420,
            "unidades_defectuosas": 18, "tiempo_ciclo_ideal": 0.9}

def cargar_oee():
    titulo("📥 DATOS DE MÁQUINA / LÍNEA (OEE)")
    if preguntar_si_no("¿Usar datos de ejemplo de máquina?"):
        print("✅ Datos de máquina de ejemplo cargados.")
        return datos_oee_ejemplo()
    plan = pedir_float("Tiempo planificado (min)", minimo=1)
    paros = pedir_float("Tiempo total de paros (min)", minimo=0)
    while paros >= plan:
        print("   ❌ Los paros deben ser menores al tiempo planificado.")
        paros = pedir_float("Tiempo total de paros (min)", minimo=0)
    total = pedir_float("Total de unidades producidas", minimo=1)
    defect = pedir_float("Unidades defectuosas", minimo=0)
    while defect > total:
        print("   ❌ Los defectos no pueden superar la producción total.")
        defect = pedir_float("Unidades defectuosas", minimo=0)
    ciclo = pedir_float("Tiempo de ciclo ideal (min/unidad)", minimo=0.001)
    return {"tiempo_planificado": plan, "tiempo_paros": paros, "total_producido": total,
            "unidades_defectuosas": defect, "tiempo_ciclo_ideal": ciclo}

# ==========================================================================
# 3. MÓDULO 3 (base): PROVEEDORES Y RIESGOS
# ==========================================================================
def mapa_material_proveedor(prov):
    """Devuelve {material: fila_proveedor}. Si hay varios, prevalece el mejor calificado."""
    mapa = {}
    for _, p in prov.iterrows():
        for m in _lista_materiales(p["Materiales_que_Suministra"]):
            if m not in mapa or p["Calificacion"] > mapa[m]["Calificacion"]:
                mapa[m] = p.to_dict()
    return mapa

def analizar_proveedores(prov, verbose=True):
    """Clasifica el riesgo de cada proveedor y propone acciones."""
    df = prov.copy()

    def riesgo(r):
        if r["Porcentaje_Entregas_A_Tiempo"] < 0.80 or r["Desviacion_Lead_Time"] > 4 or r["Calificacion"] < 7:
            return "Alto"
        if r["Porcentaje_Entregas_A_Tiempo"] < 0.90 or r["Desviacion_Lead_Time"] > 2.5 or r["Calificacion"] < 8:
            return "Medio"
        return "Bajo"

    acciones = {
        "Alto": "Buscar proveedor alterno, negociar penalizaciones por retraso y elevar el stock de seguridad.",
        "Medio": "Seguimiento semanal de pedidos, acordar fechas compromiso y revisar plan de mejora.",
        "Bajo": "Mantener relación; candidato a proveedor preferente / contratos de largo plazo.",
    }
    df["Riesgo"] = df.apply(riesgo, axis=1)
    df["Lead_Time_Efectivo"] = (df["Lead_Time_Promedio_Dias"]
                                + (1 - df["Porcentaje_Entregas_A_Tiempo"]) * df["Retraso_Promedio_Dias"]).round(1)
    df["CV_Lead_Time"] = (df["Desviacion_Lead_Time"] / df["Lead_Time_Promedio_Dias"]).round(2)
    df["Accion_Recomendada"] = df["Riesgo"].map(acciones)

    if verbose:
        titulo("🚚 MÓDULO 3 · PROVEEDORES Y RIESGOS")
        print("Lead Time efectivo = LT promedio + (1 − %a tiempo) × retraso promedio.\n")
        mostrar(df[["Proveedor", "Lead_Time_Promedio_Dias", "Lead_Time_Efectivo", "Desviacion_Lead_Time",
                    "Porcentaje_Entregas_A_Tiempo", "Retraso_Promedio_Dias", "Calificacion", "Riesgo"]])
        print("\n🔎 Interpretación y acciones:")
        for _, p in df.sort_values("Riesgo", key=lambda s: s.map({"Alto": 0, "Medio": 1, "Bajo": 2})).iterrows():
            icono = {"Alto": "🔴", "Medio": "🟡", "Bajo": "🟢"}[p["Riesgo"]]
            print(f" {icono} {p['Proveedor']} (riesgo {p['Riesgo']}): "
                  f"{p['Porcentaje_Entregas_A_Tiempo']*100:.0f}% a tiempo, σ={p['Desviacion_Lead_Time']} d. → {p['Accion_Recomendada']}")
    return df

# ==========================================================================
# 4. MÓDULO 1: GESTIÓN DE INVENTARIO AVANZADA
# ==========================================================================
def analizar_inventario(inv, prov, verbose=True):
    """EOQ, ROP, stock de seguridad (demanda + lead time), ABC, cobertura, alertas y costos."""
    df = inv.copy()
    if "Costo_Faltante_Unit" not in df.columns:
        df["Costo_Faltante_Unit"] = (df["Costo_Unitario"] * FACTOR_COSTO_FALTANTE).round(2)

    d = df["Demanda_Diaria_Prom"].astype(float)
    H = df["Costo_Mant_Unit_Anual"].astype(float)
    df["Demanda_Anual"] = d * DIAS_ANIO

    # --- EOQ = sqrt(2·D·S / H)
    df["EOQ"] = np.sqrt(2 * df["Demanda_Anual"] * df["Costo_Pedido"] / H.replace(0, np.nan)).fillna(0).round(0)

    # --- Datos del proveedor asociado a cada material
    mapa = mapa_material_proveedor(prov)
    df["Proveedor"] = df["Material"].map(lambda m: mapa[m]["Proveedor"] if m in mapa else "Sin asignar")
    df["Sigma_LT"] = df["Material"].map(lambda m: mapa[m]["Desviacion_Lead_Time"] if m in mapa else 0.0)
    df["Ajuste_LT"] = df["Material"].map(
        lambda m: (1 - mapa[m]["Porcentaje_Entregas_A_Tiempo"]) * mapa[m]["Retraso_Promedio_Dias"] if m in mapa else 0.0)
    df["LT_Efectivo"] = (df["Lead_Time_Dias"] + df["Ajuste_LT"]).round(1)

    # --- Stock de seguridad: SS = Z · sqrt( LT·σd² + d²·σLT² )
    df["Z"] = norm.ppf(df["Nivel_Servicio_Deseado"])
    sigma_ltd = np.sqrt(df["LT_Efectivo"] * df["Desv_Estandar_Demanda"] ** 2 + d ** 2 * df["Sigma_LT"] ** 2)
    df["Sigma_Demanda_LT"] = sigma_ltd
    df["Stock_Seguridad"] = (df["Z"] * sigma_ltd).round(1)

    # --- Punto de reorden: ROP = d·LT_efectivo + SS
    df["ROP"] = (d * df["LT_Efectivo"] + df["Stock_Seguridad"]).round(1)

    # --- Cobertura
    df["Dias_Cobertura"] = np.where(d > 0, df["Stock_Actual"] / d.replace(0, np.nan), 999).round(1)

    # --- Alertas de resurtido
    def estado(r):
        if r["Stock_Actual"] <= r["Stock_Seguridad"]:
            return "CRÍTICO"
        if r["Stock_Actual"] <= r["ROP"]:
            return "RESURTIR"
        if r["Dias_Cobertura"] <= r["LT_Efectivo"] * 1.5:
            return "VIGILAR"
        return "OK"
    df["Alerta"] = df.apply(estado, axis=1)
    df["Cant_Sugerida"] = np.where(df["Alerta"].isin(["CRÍTICO", "RESURTIR"]),
                                   np.maximum(df["EOQ"], df["ROP"] + df["EOQ"] - df["Stock_Actual"]), 0).round(0)

    # --- Costos anuales (pedido + mantenimiento; el SS se mantiene siempre en inventario)
    ped_anuales = np.where(df["EOQ"] > 0, df["Demanda_Anual"] / df["EOQ"].replace(0, np.nan), 0)
    df["Pedidos_Anuales"] = np.round(ped_anuales, 1)
    df["Costo_Pedido_Anual"] = (ped_anuales * df["Costo_Pedido"]).round(0)
    df["Costo_Mant_Anual"] = ((df["EOQ"] / 2 + df["Stock_Seguridad"]) * H).round(0)
    df["Costo_Total_Inv"] = df["Costo_Pedido_Anual"] + df["Costo_Mant_Anual"]

    # --- Análisis ABC (por valor anual de consumo)
    df["Valor_Anual"] = (df["Demanda_Anual"] * df["Costo_Unitario"]).round(0)
    orden = df.sort_values("Valor_Anual", ascending=False)
    pct = orden["Valor_Anual"] / orden["Valor_Anual"].sum() * 100
    acum = pct.cumsum()
    clase = np.where(acum - pct < 80, "A", np.where(acum - pct < 95, "B", "C"))
    df.loc[orden.index, "Pct_Valor"] = pct.round(1)
    df.loc[orden.index, "Pct_Acum"] = acum.round(1)
    df.loc[orden.index, "Clase_ABC"] = clase

    if verbose:
        titulo("📦 MÓDULO 1 · GESTIÓN DE INVENTARIO AVANZADA")
        print("EOQ = √(2·D·S/H) | SS = Z·√(LT·σd² + d²·σLT²) | ROP = d·LT_efectivo + SS\n"
              "(el LT_efectivo y σLT provienen del proveedor asignado a cada material)\n")
        mostrar(df[["Material", "Proveedor", "Stock_Actual", "LT_Efectivo", "Stock_Seguridad", "ROP", "EOQ",
                    "Dias_Cobertura", "Alerta", "Cant_Sugerida"]])
        print("\n💰 Costo anual de inventario (pedido + mantenimiento):")
        mostrar(df[["Material", "Pedidos_Anuales", "Costo_Pedido_Anual", "Costo_Mant_Anual", "Costo_Total_Inv"]])
        print(f"   ➜ Costo total anual de gestión de inventario: ${df['Costo_Total_Inv'].sum():,.0f}")
        print("\n🅰️ Clasificación ABC:")
        mostrar(df.sort_values("Valor_Anual", ascending=False)[["Material", "Valor_Anual", "Pct_Valor", "Pct_Acum", "Clase_ABC"]])
        print("\n🔎 Interpretación:")
        for _, r in df[df["Alerta"] == "CRÍTICO"].iterrows():
            print(f" 🔴 {r['Material']}: stock ({r['Stock_Actual']:.0f}) ya está por debajo del stock de seguridad "
                  f"({r['Stock_Seguridad']:.0f}). Ordenar ≈{r['Cant_Sugerida']:.0f} u. a {r['Proveedor']} HOY.")
        for _, r in df[df["Alerta"] == "RESURTIR"].iterrows():
            print(f" 🟠 {r['Material']}: alcanzó el punto de reorden ({r['ROP']:.0f}). Ordenar ≈{r['Cant_Sugerida']:.0f} u.")
        for _, r in df[df["Alerta"] == "VIGILAR"].iterrows():
            print(f" 🟡 {r['Material']}: cobertura de {r['Dias_Cobertura']:.0f} días, cercana al lead time efectivo ({r['LT_Efectivo']:.0f} d).")
        if (df["Alerta"] == "OK").all():
            print(" 🟢 Todos los materiales están en nivel adecuado.")
        a = df[df["Clase_ABC"] == "A"]["Material"].tolist()
        print(f" ℹ️ Clase A (≈80% del valor): {', '.join(a)} → control estricto y revisión frecuente.")
    return df

# ==========================================================================
# 5. MÓDULO 2: GESTIÓN DE PEDIDOS DE CLIENTES
# ==========================================================================
def analizar_pedidos(pedidos, inv, retrasos=None, verbose=True):
    """
    Urgencia, reparto de stock por prioridad y riesgo de incumplimiento.
    retrasos: dict opcional {material: días_extra} (se usa en el What-If).
    """
    retrasos = retrasos or {}
    hoy = datetime.now().date()
    ped = pedidos.copy()
    ped["Fecha_Pedido"] = ped["Fecha_Pedido"].apply(_a_fecha)
    ped["Fecha_Requerida"] = ped["Fecha_Requerida"].apply(_a_fecha)
    ped["Dias_Restantes"] = ped["Fecha_Requerida"].apply(lambda f: (f - hoy).days)
    ped["Urgencia"] = ped["Dias_Restantes"].apply(
        lambda x: "Urgente" if x <= UMBRAL_URGENTE else ("Próximo" if x <= UMBRAL_PROXIMO else "Normal"))

    # El stock se reparte primero a pedidos de mayor prioridad y fecha más cercana
    ped["_prio"] = ped["Prioridad"].map({"Alta": 0, "Media": 1, "Baja": 2}).fillna(1)
    orden = ped.sort_values(["_prio", "Fecha_Requerida"]).index

    inv_i = inv.set_index("Material")
    stock_rest = inv_i["Stock_Actual"].to_dict()
    for col in ("Stock_Asignado", "Faltante", "Holgura_Dias"):
        ped[col] = np.nan
    ped["Estado_Cumplimiento"] = "Sin datos"

    for idx in orden:
        r = ped.loc[idx]
        m = r["Material"]
        if m not in inv_i.index:
            continue
        asignado = min(stock_rest[m], r["Cantidad"])
        stock_rest[m] -= asignado
        falt = r["Cantidad"] - asignado
        ped.at[idx, "Stock_Asignado"] = asignado
        ped.at[idx, "Faltante"] = falt
        if falt <= 0:
            ped.at[idx, "Estado_Cumplimiento"] = "Cubierto"
        else:
            # Si se ordena HOY, ¿llega antes de la fecha requerida (escenario pesimista)?
            lt_pes = inv_i.loc[m, "LT_Efectivo"] + retrasos.get(m, 0) + Z_RIESGO_LT * inv_i.loc[m, "Sigma_LT"]
            holgura = r["Dias_Restantes"] - lt_pes
            ped.at[idx, "Holgura_Dias"] = round(holgura, 1)
            ped.at[idx, "Estado_Cumplimiento"] = "Cubrible con reposición" if holgura >= 0 else "EN RIESGO"

    ped = ped.drop(columns="_prio").sort_values("Dias_Restantes")

    if verbose:
        titulo("📋 MÓDULO 2 · PEDIDOS DE CLIENTES")
        print("Reparto de stock por prioridad (Alta > Media > Baja) y fecha requerida.\n"
              "'Holgura' = días restantes − lead time pesimista (LT + 1.645σ) del proveedor.\n")
        mostrar(ped[["Cliente", "Material", "Cantidad", "Fecha_Requerida", "Dias_Restantes", "Prioridad",
                     "Urgencia", "Stock_Asignado", "Faltante", "Holgura_Dias", "Estado_Cumplimiento"]])
        print("\n🔎 Interpretación:")
        riesgo = ped[ped["Estado_Cumplimiento"] == "EN RIESGO"]
        reposic = ped[ped["Estado_Cumplimiento"] == "Cubrible con reposición"]
        for _, r in riesgo.iterrows():
            print(f" 🔴 {r['Cliente']} – {r['Material']}: faltan {r['Faltante']:.0f} u. y el proveedor no alcanza a entregar en "
                  f"{r['Dias_Restantes']} días (holgura {r['Holgura_Dias']} d). Negociar entrega parcial / expedite / alterno.")
        for _, r in reposic.iterrows():
            print(f" 🟡 {r['Cliente']} – {r['Material']}: faltan {r['Faltante']:.0f} u.; cumplible SOLO si se ordena hoy (holgura {r['Holgura_Dias']} d).")
        if riesgo.empty and reposic.empty:
            print(" 🟢 Todos los pedidos están cubiertos con el stock actual.")
    return ped

# ==========================================================================
# 6. MÓDULO 4: OEE
# ==========================================================================
def calcular_oee(datos, verbose=True):
    plan = datos["tiempo_planificado"]
    paros = datos["tiempo_paros"]
    total = datos["total_producido"]
    defect = datos["unidades_defectuosas"]
    ciclo = datos["tiempo_ciclo_ideal"]
    t_op = plan - paros
    buenas = total - defect

    disp = t_op / plan
    rend = (ciclo * total) / t_op if t_op > 0 else 0
    cal = buenas / total if total > 0 else 0
    oee = disp * rend * cal

    # Pérdidas expresadas en minutos de tiempo planificado
    p_disp = paros
    p_rend = t_op - ciclo * total
    p_cal = ciclo * defect
    perdidas = {"Disponibilidad (paros)": p_disp, "Rendimiento (velocidad)": max(p_rend, 0),
                "Calidad (defectos)": p_cal}
    principal = max(perdidas, key=perdidas.get)

    if oee >= 0.85:
        interp = "Excelente (World Class)"
    elif oee >= 0.60:
        interp = "Aceptable / Bueno"
    elif oee >= 0.40:
        interp = "Regular - hay oportunidades claras de mejora"
    else:
        interp = "Bajo - requiere atención prioritaria"

    consejos = {
        "Disponibilidad (paros)": "Atacar paros: SMED para reducir setups, mantenimiento preventivo/TPM y análisis de causas de paro.",
        "Rendimiento (velocidad)": "Atacar microparos y velocidad reducida: estandarizar parámetros, balancear la línea y capacitar operadores.",
        "Calidad (defectos)": "Atacar defectos: control estadístico de proceso, poka-yoke y análisis de causa raíz (5 porqués / Pareto).",
    }

    res = {"Disponibilidad": disp, "Rendimiento": rend, "Calidad": cal, "OEE": oee, "Interpretacion": interp,
           "Perdidas_min": perdidas, "Principal_Perdida": principal, "Consejo": consejos[principal], "datos": datos}

    if verbose:
        titulo("⚙️ MÓDULO 4 · OEE (Overall Equipment Effectiveness)")
        print("OEE = Disponibilidad × Rendimiento × Calidad\n")
        print(f" 📌 Disponibilidad: {disp*100:5.1f}%   (operó {t_op:.0f} de {plan:.0f} min)")
        print(f" 📌 Rendimiento:    {rend*100:5.1f}%   (ciclo ideal {ciclo} min/u)")
        print(f" 📌 Calidad:        {cal*100:5.1f}%   ({buenas:.0f} buenas de {total:.0f})")
        print(f" 🎯 OEE TOTAL:      {oee*100:5.1f}%  → {interp}")
        if rend > 1:
            print(" ⚠️ Rendimiento > 100%: revisa el tiempo de ciclo ideal o la producción registrada.")
        print("\n📉 Análisis de pérdidas (minutos del tiempo planificado):")
        for k, v in perdidas.items():
            print(f"   • {k:26s}: {v:6.1f} min ({v/plan*100:4.1f}% del tiempo planificado)")
        print(f"\n 🔎 Principal área de mejora: {principal}.\n    {consejos[principal]}")
        print(f"    Meta World Class: 85%. Brecha actual: {max(0, 0.85-oee)*100:.1f} puntos.")
    return res

# ==========================================================================
# 7. MÓDULO 5: COSTOS Y ESCENARIOS WHAT-IF
# ==========================================================================
def _perdida_normal(z):
    """Función de pérdida estándar normal L(z) = φ(z) − z·(1 − Φ(z))."""
    return norm.pdf(z) - z * (1 - norm.cdf(z))

def simular_retraso(inv, prov, pedidos, proveedor, dias_extra, verbose=True):
    """
    ¿Qué pasa si 'proveedor' se atrasa 'dias_extra' días más?
    Mantiene la política actual (ROP) y mide el efecto en nivel de servicio, costo de faltantes y pedidos.
    """
    mapa = mapa_material_proveedor(prov)
    mats = [m for m, p in mapa.items() if p["Proveedor"] == proveedor]
    sub = inv[inv["Material"].isin(mats)].copy()
    if sub.empty:
        return None

    d, sd, slt = sub["Demanda_Diaria_Prom"], sub["Desv_Estandar_Demanda"], sub["Sigma_LT"]
    lt0 = sub["LT_Efectivo"]
    lt1 = lt0 + dias_extra
    sig0 = np.maximum(np.sqrt(lt0 * sd ** 2 + d ** 2 * slt ** 2), 1e-9)
    sig1 = np.maximum(np.sqrt(lt1 * sd ** 2 + d ** 2 * slt ** 2), 1e-9)

    ns0 = norm.cdf((sub["ROP"] - d * lt0) / sig0)
    z1 = (sub["ROP"] - d * lt1) / sig1
    ns1 = norm.cdf(z1)

    ciclos = np.where(sub["EOQ"] > 0, sub["Demanda_Anual"] / sub["EOQ"].replace(0, np.nan), 0)
    falt_ciclo = sig1 * _perdida_normal(z1)
    costo_falt = falt_ciclo * ciclos * sub["Costo_Faltante_Unit"]

    ss_nec = sub["Z"] * sig1
    costo_extra_mant = (ss_nec - sub["Stock_Seguridad"]) * sub["Costo_Mant_Unit_Anual"]

    out = pd.DataFrame({
        "Material": sub["Material"], "LT_Actual": lt0.round(1), "LT_Con_Retraso": lt1.round(1),
        "NS_Actual_%": (ns0 * 100).round(1), "NS_Con_Retraso_%": (ns1 * 100).round(1),
        "Faltante_Esp_por_Ciclo": falt_ciclo.round(1), "Costo_Faltantes_Anual": costo_falt.round(0),
        "SS_Necesario": ss_nec.round(1), "Costo_Extra_Mantener_SS": costo_extra_mant.round(0),
        "Se_Agota_Antes": sub["Dias_Cobertura"] < lt1,
    })

    # Impacto en pedidos de clientes
    base = analizar_pedidos(pedidos, inv, verbose=False)
    esc = analizar_pedidos(pedidos, inv, retrasos={m: dias_extra for m in mats}, verbose=False)
    riesgo_base = int((base["Estado_Cumplimiento"] == "EN RIESGO").sum())
    riesgo_esc = int((esc["Estado_Cumplimiento"] == "EN RIESGO").sum())
    nuevos = esc[(esc["Estado_Cumplimiento"] == "EN RIESGO") &
                 (base.loc[esc.index, "Estado_Cumplimiento"] != "EN RIESGO")]

    if verbose:
        print(f"\n🧪 ESCENARIO: {proveedor} se retrasa +{dias_extra} días  (materiales: {', '.join(mats)})")
        mostrar(out)
        print(f" 🔎 Nivel de servicio promedio: {out['NS_Actual_%'].mean():.1f}% → {out['NS_Con_Retraso_%'].mean():.1f}%")
        print(f" 💸 Costo esperado de faltantes: ${out['Costo_Faltantes_Anual'].sum():,.0f}/año  |  "
              f"Costo de compensar con más stock de seguridad: ${out['Costo_Extra_Mantener_SS'].sum():,.0f}/año")
        mejor = "mantener más stock de seguridad" if out["Costo_Extra_Mantener_SS"].sum() < out["Costo_Faltantes_Anual"].sum() \
            else "asumir el riesgo de faltantes"
        print(f" ✅ Decisión económica sugerida: {mejor}.")
        print(f" 📋 Pedidos en riesgo: {riesgo_base} → {riesgo_esc}")
        for _, r in nuevos.iterrows():
            print(f"    • Nuevo pedido en riesgo: {r['Cliente']} – {r['Material']} (holgura {r['Holgura_Dias']} d)")
    return {"tabla": out, "riesgo_base": riesgo_base, "riesgo_escenario": riesgo_esc,
            "nuevos_en_riesgo": nuevos, "proveedor": proveedor, "dias_extra": dias_extra, "materiales": mats}

def sensibilidad_retraso(inv, prov, pedidos, proveedor, retrasos=(0, 2, 4, 7, 10, 14)):
    filas = []
    for r in retrasos:
        s = simular_retraso(inv, prov, pedidos, proveedor, r, verbose=False)
        if s is None:
            return None
        filas.append({"Retraso_Dias": r, "NS_Promedio_%": s["tabla"]["NS_Con_Retraso_%"].mean(),
                      "Pedidos_En_Riesgo": s["riesgo_escenario"],
                      "Costo_Faltantes": s["tabla"]["Costo_Faltantes_Anual"].sum()})
    return pd.DataFrame(filas)

def curva_costos(inv, material):
    """Costo de mantener (stock de seguridad) vs costo de faltantes según nivel de servicio."""
    r = inv[inv["Material"] == material].iloc[0]
    ciclos = r["Demanda_Anual"] / r["EOQ"] if r["EOQ"] > 0 else 0
    niveles = np.linspace(0.50, 0.999, 80)
    z = norm.ppf(niveles)
    sig = r["Sigma_Demanda_LT"]
    mant = z * sig * r["Costo_Mant_Unit_Anual"]
    falt = sig * _perdida_normal(z) * ciclos * r["Costo_Faltante_Unit"]
    df = pd.DataFrame({"Nivel_Servicio": niveles, "Costo_Mantener": mant, "Costo_Faltantes": falt})
    df["Costo_Total"] = df["Costo_Mantener"] + df["Costo_Faltantes"]
    return df

def modulo_whatif(inv, prov, pedidos):
    titulo("🧪 MÓDULO 5 · COSTOS Y ESCENARIOS WHAT-IF")
    print("Compara el costo de mantener inventario vs el costo de faltantes y simula retrasos de proveedores.\n")
    # --- Costo óptimo de servicio (material Clase A de mayor valor)
    mat = inv.sort_values("Valor_Anual", ascending=False).iloc[0]["Material"]
    curva = curva_costos(inv, mat)
    opt = curva.loc[curva["Costo_Total"].idxmin()]
    ns_act = inv.loc[inv["Material"] == mat, "Nivel_Servicio_Deseado"].iloc[0]
    print(f"💡 Trade-off para '{mat}' (material de mayor valor):")
    print(f"   • Nivel de servicio económicamente óptimo ≈ {opt['Nivel_Servicio']*100:.1f}% "
          f"(costo total ${opt['Costo_Total']:,.0f}/año). Nivel objetivo actual: {ns_act*100:.1f}%.")
    print("   • Más servicio = más stock de seguridad (cuesta mantener); menos servicio = más faltantes (cuesta no vender).")
    print("   • Si el cliente exige un nivel mayor al óptimo, el sobrecosto es el precio del servicio.")

    # --- Escenario sobre el proveedor de mayor riesgo
    pr = analizar_proveedores(prov, verbose=False)
    orden = pr.assign(_o=pr["Riesgo"].map({"Alto": 0, "Medio": 1, "Bajo": 2})).sort_values(["_o", "Porcentaje_Entregas_A_Tiempo"])
    defecto = orden.iloc[0]["Proveedor"]
    print(f"\nProveedores disponibles: {', '.join(pr['Proveedor'])}")
    proveedor = pedir_texto("Proveedor a simular", defecto)
    if proveedor not in pr["Proveedor"].values:
        print(f"⚠️ '{proveedor}' no existe; se usará '{defecto}'.")
        proveedor = defecto
    dias = pedir_float("Días de retraso adicionales a simular", 7, 0) if not MODO_DEMO else 7.0
    esc = simular_retraso(inv, prov, pedidos, proveedor, dias)
    sens = sensibilidad_retraso(inv, prov, pedidos, proveedor)
    if sens is not None:
        print("\n📈 Sensibilidad al retraso:")
        mostrar(sens.round(1))
    return {"material_curva": mat, "curva": curva, "optimo": opt, "escenario": esc, "sensibilidad": sens}

# ==========================================================================
# 8. EXTRA: OPTIMIZACIÓN DEL MIX DE PRODUCCIÓN CON PuLP
# ==========================================================================
PRODUCTOS_EJEMPLO = {
    # nombre: margen unitario, minutos de ciclo ideal, demanda máxima del horizonte, materiales por unidad
    "Bomba Hidráulica": {"margen": 1800, "min": 45, "dem_max": 10, "mat": {"Motor 1.5HP": 1, "Válvula": 2, "Sensor": 2, "Cable 3x12": 3}},
    "Panel de Control": {"margen": 420, "min": 25, "dem_max": 30, "mat": {"Sensor": 4, "Cable 3x12": 10, "Placa Acero": 1}},
    "Soporte Reforzado": {"margen": 150, "min": 12, "dem_max": 60, "mat": {"Placa Acero": 2, "Tornillo M8": 12, "Rodamiento": 2}},
}

def optimizar_produccion(inv, oee, productos=None, dias=HORIZONTE_PROD_DIAS, verbose=True):
    """
    Programación lineal entera: maximiza margen total sujeto a capacidad de máquina
    (ajustada por OEE) y disponibilidad de materiales en inventario.
    Edita el diccionario PRODUCTOS_EJEMPLO para usar tus propios productos.
    """
    productos = productos or PRODUCTOS_EJEMPLO
    stock = inv.set_index("Material")["Stock_Actual"].to_dict()
    validos = {}
    for nombre, p in productos.items():
        faltan = [m for m in p["mat"] if m not in stock]
        if faltan:
            if verbose:
                print(f"   ⚠️ '{nombre}' omitido: materiales no encontrados en inventario: {', '.join(faltan)}")
        else:
            validos[nombre] = p
    if not validos:
        if verbose:
            print("   ⚠️ Ningún producto compatible con tu inventario. Edita PRODUCTOS_EJEMPLO.")
        return None

    cap_min = oee["datos"]["tiempo_planificado"] * dias * oee["OEE"]   # minutos ideales útiles
    prob = LpProblem("Mix_Produccion", LpMaximize)
    x = {n: LpVariable(f"x_{i}", 0, p["dem_max"], cat="Integer") for i, (n, p) in enumerate(validos.items())}
    prob += lpSum(p["margen"] * x[n] for n, p in validos.items()), "Margen_Total"
    prob += lpSum(p["min"] * x[n] for n, p in validos.items()) <= cap_min, "Capacidad_Maquina"
    mats_usados = sorted({m for p in validos.values() for m in p["mat"]})
    for k, m in enumerate(mats_usados):
        prob += lpSum(p["mat"].get(m, 0) * x[n] for n, p in validos.items()) <= stock[m], f"Stock_{k}"
    prob.solve(PULP_CBC_CMD(msg=0))

    plan = pd.DataFrame([{"Producto": n, "Unidades": int(round(x[n].value() or 0)), "Margen_Unit": p["margen"],
                          "Margen_Total": int(round(x[n].value() or 0)) * p["margen"], "Demanda_Max": p["dem_max"]}
                         for n, p in validos.items()])
    uso_cap = sum(validos[n]["min"] * plan.set_index("Producto").loc[n, "Unidades"] for n in validos)
    util = [{"Recurso": "Capacidad de máquina (min)", "Usado": round(uso_cap, 1), "Disponible": round(cap_min, 1)}]
    for m in mats_usados:
        u = sum(validos[n]["mat"].get(m, 0) * plan.set_index("Producto").loc[n, "Unidades"] for n in validos)
        util.append({"Recurso": f"Material: {m}", "Usado": u, "Disponible": stock[m]})
    util = pd.DataFrame(util)
    util["Utilizacion_%"] = (util["Usado"] / util["Disponible"].replace(0, np.nan) * 100).round(1)
    cuellos = util[util["Utilizacion_%"] >= 99]["Recurso"].tolist()

    res = {"estado": LpStatus[prob.status], "plan": plan, "utilizacion": util,
           "margen_total": value(prob.objective), "cuellos": cuellos, "capacidad_min": cap_min}
    if verbose:
        titulo("🏭 EXTRA · OPTIMIZACIÓN DEL MIX DE PRODUCCIÓN (PuLP)")
        print(f"Horizonte: {dias} días. Capacidad útil = tiempo planificado × días × OEE = {cap_min:,.0f} min.\n")
        print(f"Estado del solver: {res['estado']}")
        mostrar(plan)
        print(f"💰 Margen total óptimo: ${res['margen_total']:,.0f}")
        mostrar(util)
        if cuellos:
            print(f"🔎 Cuellos de botella (≥99% de uso): {', '.join(cuellos)}. "
                  "Aumentar estos recursos es lo que más margen adicional generaría.")
    return res

# ==========================================================================
# 9. MÓDULO 6: DASHBOARD VISUAL
# ==========================================================================
def dashboard(inv, ped, prov_a, oee, titulo_fig="SmartIE · Dashboard Ejecutivo"):
    titulo("📊 MÓDULO 6 · DASHBOARD VISUAL")
    fig, ax = plt.subplots(3, 2, figsize=(15, 16))
    fig.suptitle(titulo_fig, fontsize=17, fontweight="bold", y=0.995)

    # 1) Stock vs ROP
    a = ax[0, 0]
    x = np.arange(len(inv)); w = 0.27
    a.bar(x - w, inv["Stock_Actual"], w, label="Stock actual", color="#3498db")
    a.bar(x, inv["ROP"], w, label="Punto de reorden", color="#e74c3c")
    a.bar(x + w, inv["Stock_Seguridad"], w, label="Stock de seguridad", color="#f39c12")
    a.set_xticks(x); a.set_xticklabels(inv["Material"], rotation=40, ha="right", fontsize=8)
    a.set_title("1. Stock actual vs Punto de Reorden"); a.set_ylabel("Unidades"); a.legend(fontsize=8)

    # 2) OEE
    a = ax[0, 1]
    nombres = ["Disponibilidad", "Rendimiento", "Calidad", "OEE"]
    vals = [oee[k] * 100 for k in nombres]
    barras = a.bar(nombres, vals, color=["#3498db", "#2ecc71", "#f1c40f", "#9b59b6"])
    a.axhline(85, color="green", ls="--", alpha=.7, label="World Class 85%")
    a.axhline(60, color="orange", ls="--", alpha=.7, label="Mínimo aceptable 60%")
    a.set_ylim(0, 110); a.set_ylabel("%"); a.set_title("2. OEE y sus componentes"); a.legend(fontsize=8)
    for b, v in zip(barras, vals):
        a.text(b.get_x() + b.get_width() / 2, v + 1.5, f"{v:.1f}%", ha="center", fontsize=9)

    # 3) Confiabilidad proveedores
    a = ax[1, 0]
    pc = prov_a.sort_values("Porcentaje_Entregas_A_Tiempo")
    cols = pc["Riesgo"].map({"Alto": "#e74c3c", "Medio": "#f39c12", "Bajo": "#2ecc71"})
    a.barh(pc["Proveedor"], pc["Porcentaje_Entregas_A_Tiempo"] * 100, color=cols)
    a.axvline(90, color="green", ls="--", alpha=.7, label="Meta 90%")
    a.axvline(80, color="orange", ls="--", alpha=.7, label="Mínimo 80%")
    a.set_xlim(50, 100); a.set_xlabel("% entregas a tiempo"); a.set_title("3. Confiabilidad de proveedores (color = riesgo)")
    a.legend(fontsize=8)

    # 4) Días de cobertura vs LT efectivo
    a = ax[1, 1]
    ic = inv.sort_values("Dias_Cobertura")
    colores = ic["Alerta"].map({"CRÍTICO": "#c0392b", "RESURTIR": "#e67e22", "VIGILAR": "#f1c40f", "OK": "#2ecc71"})
    a.barh(ic["Material"], ic["Dias_Cobertura"].clip(upper=200), color=colores)
    a.scatter(ic["LT_Efectivo"], ic["Material"], color="black", marker="|", s=250, label="Lead time efectivo", zorder=3)
    a.set_xlabel("Días de cobertura"); a.set_title("4. Cobertura de inventario (color = alerta)"); a.legend(fontsize=8)

    # 5) Pareto ABC
    a = ax[2, 0]
    ia = inv.sort_values("Valor_Anual", ascending=False)
    cmap = ia["Clase_ABC"].map({"A": "#8e44ad", "B": "#3498db", "C": "#95a5a6"})
    a.bar(ia["Material"], ia["Valor_Anual"], color=cmap)
    a.set_xticklabels(ia["Material"], rotation=40, ha="right", fontsize=8)
    a.set_ylabel("Valor anual de consumo ($)")
    a2 = a.twinx()
    a2.plot(ia["Material"], ia["Pct_Acum"], color="#e74c3c", marker="o")
    a2.axhline(80, color="gray", ls="--", alpha=.6); a2.axhline(95, color="gray", ls=":", alpha=.6)
    a2.set_ylim(0, 105); a2.set_ylabel("% acumulado"); a2.grid(False)
    a.set_title("5. Análisis ABC (Pareto de valor)")

    # 6) Estado de pedidos
    a = ax[2, 1]
    if ped is not None and len(ped):
        sns.countplot(data=ped, x="Urgencia", hue="Estado_Cumplimiento", order=["Urgente", "Próximo", "Normal"],
                      palette=COL_ESTADO, ax=a)
        a.set_ylabel("Número de pedidos"); a.legend(title="", fontsize=8)
        a.yaxis.get_major_locator().set_params(integer=True)
    a.set_title("6. Estado de pedidos de clientes por urgencia")

    plt.tight_layout(rect=[0, 0, 1, 0.985])
    plt.show()

def dashboard_escenarios(whatif):
    if whatif is None:
        return
    fig, ax = plt.subplots(1, 3, figsize=(17, 4.8))
    c, o = whatif["curva"], whatif["optimo"]
    ax[0].plot(c["Nivel_Servicio"] * 100, c["Costo_Mantener"], label="Mantener (SS)", color="#3498db")
    ax[0].plot(c["Nivel_Servicio"] * 100, c["Costo_Faltantes"], label="Faltantes", color="#e74c3c")
    ax[0].plot(c["Nivel_Servicio"] * 100, c["Costo_Total"], label="Total", color="black", lw=2)
    ax[0].axvline(o["Nivel_Servicio"] * 100, color="green", ls="--", label=f"Óptimo {o['Nivel_Servicio']*100:.1f}%")
    ax[0].set_xlabel("Nivel de servicio (%)"); ax[0].set_ylabel("$ / año")
    ax[0].set_title(f"Mantener vs Faltantes · {whatif['material_curva']}"); ax[0].legend(fontsize=8)

    s = whatif["sensibilidad"]
    if s is not None:
        ax[1].plot(s["Retraso_Dias"], s["NS_Promedio_%"], marker="o", color="#8e44ad")
        ax[1].set_xlabel("Retraso adicional del proveedor (días)"); ax[1].set_ylabel("Nivel de servicio promedio (%)")
        ax[1].set_title(f"Impacto en nivel de servicio · {whatif['escenario']['proveedor']}")
        ax[2].bar(s["Retraso_Dias"].astype(str), s["Pedidos_En_Riesgo"], color="#e74c3c")
        ax[2].set_xlabel("Retraso adicional (días)"); ax[2].set_ylabel("Pedidos en riesgo")
        ax[2].set_title("Impacto en pedidos de clientes")
        ax[2].yaxis.get_major_locator().set_params(integer=True)
    plt.tight_layout(); plt.show()

# ==========================================================================
# 10. MÓDULO 7: RESUMEN EJECUTIVO + EXPORTACIÓN
# ==========================================================================
def resumen_ejecutivo(inv, ped, prov_a, oee, whatif=None, prod=None, empresa="Mi Empresa"):
    titulo("📝 MÓDULO 7 · RESUMEN EJECUTIVO", 72)
    print(f"Empresa: {empresa}\nFecha del análisis: {datetime.now().strftime('%d/%m/%Y %H:%M')}\n")
    acciones = []

    print("1) MATERIALES QUE REQUIEREN RESURTIDO")
    urg = inv[inv["Alerta"].isin(["CRÍTICO", "RESURTIR"])].sort_values("Alerta")
    if urg.empty:
        print("   🟢 Ninguno por ahora.")
    for _, r in urg.iterrows():
        ic = "🔴" if r["Alerta"] == "CRÍTICO" else "🟠"
        print(f"   {ic} {r['Material']} [{r['Alerta']}]: stock {r['Stock_Actual']:.0f} / ROP {r['ROP']:.0f} → "
              f"ordenar ≈{r['Cant_Sugerida']:.0f} u. a {r['Proveedor']} (clase {r['Clase_ABC']}).")
        acciones.append(f"Emitir orden de compra de {r['Cant_Sugerida']:.0f} u. de {r['Material']} a {r['Proveedor']}.")

    print("\n2) PROVEEDORES DE ALTO RIESGO")
    altos = prov_a[prov_a["Riesgo"] == "Alto"]
    if altos.empty:
        print("   🟢 Ninguno.")
    for _, p in altos.iterrows():
        print(f"   🔴 {p['Proveedor']}: {p['Porcentaje_Entregas_A_Tiempo']*100:.0f}% a tiempo, σ={p['Desviacion_Lead_Time']} d, "
              f"retraso prom. {p['Retraso_Promedio_Dias']} d.")
        acciones.append(f"Plan de contingencia con {p['Proveedor']} (alterno / penalizaciones / más stock de seguridad).")

    print("\n3) PEDIDOS DE CLIENTES EN RIESGO")
    rr = ped[ped["Estado_Cumplimiento"].isin(["EN RIESGO", "Cubrible con reposición"])]
    if rr.empty:
        print("   🟢 Todos cubiertos con el stock actual.")
    for _, r in rr.iterrows():
        ic = "🔴" if r["Estado_Cumplimiento"] == "EN RIESGO" else "🟡"
        print(f"   {ic} {r['Cliente']} – {r['Material']} ({r['Prioridad']}): faltan {r['Faltante']:.0f} u., "
              f"{r['Dias_Restantes']} d restantes → {r['Estado_Cumplimiento']}.")
        if r["Estado_Cumplimiento"] == "EN RIESGO":
            acciones.append(f"Contactar a {r['Cliente']} para renegociar/entregar parcial {r['Material']}.")

    print("\n4) EFICIENCIA DE EQUIPO (OEE)")
    print(f"   OEE = {oee['OEE']*100:.1f}% ({oee['Interpretacion']}). Principal pérdida: {oee['Principal_Perdida']}.")
    acciones.append(oee["Consejo"])

    if whatif and whatif.get("escenario"):
        e = whatif["escenario"]
        print("\n5) ESCENARIO WHAT-IF")
        print(f"   Si {e['proveedor']} se retrasa +{e['dias_extra']:.0f} d: nivel de servicio promedio "
              f"{e['tabla']['NS_Actual_%'].mean():.1f}% → {e['tabla']['NS_Con_Retraso_%'].mean():.1f}%; "
              f"pedidos en riesgo {e['riesgo_base']} → {e['riesgo_escenario']}.")
    if prod:
        print("\n6) PLAN DE PRODUCCIÓN ÓPTIMO")
        print(f"   Margen máximo: ${prod['margen_total']:,.0f}. "
              f"Cuellos de botella: {', '.join(prod['cuellos']) if prod['cuellos'] else 'ninguno'}.")

    print("\n🎯 ACCIONES PRIORITARIAS")
    for i, a in enumerate(dict.fromkeys(acciones), 1):
        print(f"   {i}. {a}")
    print("\n💵 Costo anual de gestión de inventario (pedido + mantenimiento): "
          f"${inv['Costo_Total_Inv'].sum():,.0f}")

def exportar_resultados(res, archivo="SmartIE_resultados.xlsx"):
    """Exporta las tablas principales a Excel (y descarga en Colab si es posible)."""
    hojas = {"Inventario": res["inventario"], "Pedidos": res["pedidos"], "Proveedores": res["proveedores"]}
    if res.get("produccion"):
        hojas["Plan_Produccion"] = res["produccion"]["plan"]
    if res.get("whatif") and res["whatif"].get("escenario"):
        hojas["WhatIf_Retraso"] = res["whatif"]["escenario"]["tabla"]
    hojas["OEE"] = pd.DataFrame([{k: v for k, v in res["oee"].items()
                                  if k in ("Disponibilidad", "Rendimiento", "Calidad", "OEE", "Interpretacion", "Principal_Perdida")}])
    try:
        with pd.ExcelWriter(archivo) as w:
            for n, d in hojas.items():
                d.to_excel(w, sheet_name=n, index=False)
        print(f"\n💾 Resultados exportados a '{archivo}'.")
        try:
            from google.colab import files
            files.download(archivo)
        except Exception:
            pass
    except Exception as e:
        print(f"⚠️ No se pudo crear el Excel ({e}). Los DataFrames siguen disponibles en la variable 'resultados'.")

# ==========================================================================
# 11. EJECUCIÓN PRINCIPAL
# ==========================================================================
def main():
    global MODO_DEMO
    print("🚀 SmartIE – Sistema Inteligente de Apoyo a Decisiones para Ingeniería Industrial")
    print("   Módulos: Inventario · Pedidos · Proveedores · OEE · What-If · Dashboard · Resumen\n")
    MODO_DEMO = input("¿Ejecutar TODO con datos de ejemplo (modo demo)? (s/n): ").strip().lower() == "s"
    if MODO_DEMO:
        empresa = EMPRESA_EJEMPLO
        print(f"\n🏢 CASO DE DEMOSTRACIÓN: {empresa}\n   {PERFIL_EMPRESA}")
    else:
        empresa = pedir_texto("Nombre de tu empresa", "Mi Empresa")

    # --- Carga de datos
    inventario = cargar_inventario()
    pedidos = cargar_pedidos()
    proveedores = cargar_proveedores()
    datos_oee = cargar_oee()

    # --- Análisis (orden: proveedores → inventario → pedidos → OEE)
    prov_a = analizar_proveedores(proveedores)
    inv_a = analizar_inventario(inventario, proveedores)
    ped_a = analizar_pedidos(pedidos, inv_a)
    oee = calcular_oee(datos_oee)
    whatif = modulo_whatif(inv_a, proveedores, pedidos)
    try:
        prod = optimizar_produccion(inv_a, oee)
    except Exception as e:   # el resto del sistema sigue funcionando aunque falle el solver
        print(f"⚠️ No se pudo ejecutar la optimización con PuLP ({e}).")
        prod = None

    # --- Visualización y conclusiones
    dashboard(inv_a, ped_a, prov_a, oee, titulo_fig=f"SmartIE · Dashboard Ejecutivo\n{empresa}")
    dashboard_escenarios(whatif)
    resumen_ejecutivo(inv_a, ped_a, prov_a, oee, whatif, prod, empresa)

    resultados = {"inventario": inv_a, "pedidos": ped_a, "proveedores": prov_a, "oee": oee,
                  "whatif": whatif, "produccion": prod}
    exportar_resultados(resultados)

    print("""
============================================================================
 CÓMO ADAPTAR SmartIE A DATOS REALES DE UNA EMPRESA
============================================================================
 1. Datos: exporta desde tu ERP/Excel y cárgalos con pd.read_csv() / pd.read_excel()
    (en Colab: files.upload()) respetando los nombres de columnas de cada módulo.
 2. Inventario: usa demanda diaria y su desviación de los últimos 6-12 meses
    (df.groupby(...).agg(['mean','std'])). Ajusta nivel de servicio por clase ABC.
 3. Proveedores: calcula % a tiempo, retraso promedio y σ del lead time con el
    histórico real de recepciones (fecha prometida vs. fecha real).
 4. Costos: reemplaza FACTOR_COSTO_FALTANTE por el margen/penalización real.
 5. OEE: alimenta con datos de MES/SCADA o bitácoras de turno (por máquina).
 6. Producción: edita el diccionario PRODUCTOS_EJEMPLO (margen, minutos, materiales).
 7. Umbrales: UMBRAL_URGENTE, UMBRAL_PROXIMO y reglas de riesgo se modifican arriba.
 8. Resultados: la variable 'resultados' contiene todos los DataFrames y el Excel exportado.
============================================================================""")
    return resultados

if __name__ == "__main__":
    resultados = main()
```
